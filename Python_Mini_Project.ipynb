```python
# =======================================================
# Python Mini-Project (Day 5 Assignment)
# Student Grade Management System (OOP-Based Colab-Safe Version)
# =======================================================

import csv
import os


class Student:
    """Class to represent an individual student."""

    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name
        self.grades = {}

    def add_grade(self, subject, grade):
        self.grades[subject] = float(grade)

    def calculate_average(self):
        if not self.grades:
            return 0.0
        return sum(self.grades.values()) / len(self.grades)

    def get_letter_grade(self):
        avg = self.calculate_average()
        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        else:
            return "F"

    def to_dict(self):
        return {
            "student_id": self.student_id,
            "name": self.name,
            "grades": str(self.grades),
            "average": round(self.calculate_average(), 2),
            "letter_grade": self.get_letter_grade(),
        }


class GradeManagerSystem:
    """Class to manage multiple students and system operations."""

    def __init__(self, filename="students_record.csv"):
        self.students = {}
        self.filename = filename

    def add_student(self, student_id, name):
        if student_id in self.students:
            print(f"Error: Student ID '{student_id}' already exists!")
            return
        self.students[student_id] = Student(student_id, name)
        print(f"Student '{name}' added successfully.")

    def add_grade(self, student_id, subject, grade):
        if student_id not in self.students:
            print(f"Error: Student ID '{student_id}' not found.")
            return
        if 0 <= grade <= 100:
            self.students[student_id].add_grade(subject, grade)
            print(
                f"Grade {grade} added for '{subject}' (Student ID: {student_id})."
            )
        else:
            print("Error: Grade must be between 0 and 100.")

    def view_all_students(self):
        if not self.students:
            print("\nNo student records found.")
            return

        print("\n" + "=" * 65)
        print(
            f"{'ID':<10} | {'Name':<20} | {'Average':<10} | {'Letter Grade':<12}"
        )
        print("-" * 65)
        for student in self.students.values():
            avg = round(student.calculate_average(), 2)
            letter = student.get_letter_grade()
            print(
                f"{student.student_id:<10} | {student.name:<20} | {avg:<10} | {letter:<12}"
            )
        print("=" * 65)

    def search_student(self, student_id):
        if student_id in self.students:
            student = self.students[student_id]
            print(f"\n--- Student Details (ID: {student_id}) ---")
            print(f"Name: {student.name}")
            print(f"Grades: {student.grades}")
            print(f"Average: {student.calculate_average():.2f}")
            print(f"Performance: {student.get_letter_grade()}")
        else:
            print(f"Error: Student ID '{student_id}' not found.")

    def save_to_csv(self):
        try:
            with open(
                self.filename, mode="w", newline="", encoding="utf-8"
            ) as file:
                fieldnames = [
                    "student_id",
                    "name",
                    "grades",
                    "average",
                    "letter_grade",
                ]
                writer = csv.DictWriter(file, fieldnames=fieldnames)
                writer.writeheader()
                for student in self.students.values():
                    writer.writerow(student.to_dict())
            print(f"\nRecords saved successfully to '{self.filename}'.")
        except Exception as e:
            print(f"Error saving data: {e}")


# -------------------------------------------------------
# Demonstrate System Operations (No infinite loop)
# -------------------------------------------------------
if __name__ == "__main__":
    system = GradeManagerSystem()

    # 1. Add Sample Students
    system.add_student("S101", "Alice Johnson")
    system.add_student("S102", "Bob Smith")
    system.add_student("S103", "Charlie Brown")

    # 2. Assign Subject Grades
    system.add_grade("S101", "Math", 95)
    system.add_grade("S101", "Science", 88)
    system.add_grade("S102", "Math", 72)
    system.add_grade("S102", "English", 68)
    system.add_grade("S103", "History", 82)

    # 3. View All Students
    system.view_all_students()

    # 4. Search Specific Student
    system.search_student("S101")

    # 5. Export Data
    system.save_to_csv()
```
